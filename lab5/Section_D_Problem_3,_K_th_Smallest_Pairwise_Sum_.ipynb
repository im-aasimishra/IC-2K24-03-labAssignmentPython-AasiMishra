```python
# ============================================================
# Python Lab Assignment - Lab 05
# Section B - Problem 3
# Program: Kth Smallest Pair Sum
# Name: Aasi Mishra
# Roll No: IC-2K24-03
# ============================================================

import random
import time


# ------------------------------------------------------------
# Naive Approach
# ------------------------------------------------------------

def kth_smallest_sum_naive(arr, k):
    sums = []
    n = len(arr)

    for i in range(n):
        for j in range(i + 1, n):
            sums.append(arr[i] + arr[j])

    sums.sort()
    return sums[k - 1]


# ------------------------------------------------------------
# Count Pairs with Sum Less Than or Equal to Given Value
# ------------------------------------------------------------

def count_pairs_leq(arr, v):
    n = len(arr)
    count = 0
    left, right = 0, n - 1

    while left < right:
        if arr[left] + arr[right] <= v:
            count += right - left
            left += 1
        else:
            right -= 1

    return count


# ------------------------------------------------------------
# Optimized Approach
# ------------------------------------------------------------

def kth_smallest_sum_optimized(arr, k):
    arr = sorted(arr)
    n = len(arr)

    lo = arr[0] + arr[1]
    hi = arr[n - 2] + arr[n - 1]

    while lo < hi:
        mid = (lo + hi) // 2

        if count_pairs_leq(arr, mid) >= k:
            hi = mid
        else:
            lo = mid + 1

    return lo


# ------------------------------------------------------------
# Input Generation
# ------------------------------------------------------------

random.seed(42)

N = 5000
arr = [random.randint(1, 100000) for _ in range(N)]
K = 1000000


# ------------------------------------------------------------
# Performance Testing - Naive Version
# ------------------------------------------------------------

print("Naive Version")

start = time.time()
result_naive = kth_smallest_sum_naive(arr, K)
end = time.time()

print(f"N = {N}, K = {K}, result = {result_naive}, time = {end - start:.4f} s")


# ------------------------------------------------------------
# Performance Testing - Optimized Version
# ------------------------------------------------------------

print()
print("Optimized Version")

start = time.time()
result_opt = kth_smallest_sum_optimized(arr, K)
end = time.time()

print(f"N = {N}, K = {K}, result = {result_opt}, time = {end - start:.4f} s")


# ------------------------------------------------------------
# Result Verification
# ------------------------------------------------------------

print()

if result_naive == result_opt:
    print("Both methods produced the same result: PASS")
else:
    print("Results do not match: FAIL")
```
